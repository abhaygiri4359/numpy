```python
b=[12,23,45,65,32,23]
x=np.array(b)
x=x.reshape(2,3)
print(x)
print("data type of x",x.dtype)
print("dimention of x",x.ndim)
print("shape of x",x.shape)
print("size of x",x.size)
```

```python
# to convert 2d array into one d reshape not be used to do this we have to use shape attribute
```

```python
import numpy as np

x = np.array([[12, 23, 45],
              [65, 32, 23]])

print(x.shape)    # (2, 3)

x.shape = (6,)    # 2D se 1D
print(x)          # [12 23 45 65 32 23]
print(x.ndim)     # 1
print(x.shape)    # (6,)
```
