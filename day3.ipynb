```python
#numpy.arange(value)-->0 to value-1 with default step +1
#numpy.arange(start,stop)--> from start to stop-1 with default step +1
#numpy.arange(start,stop,step)-->from start to stop with given step
```

```python
import numpy as np
a=np.arange(11)
print("content of a",a)
print("data type of a",type(a))
print("datatype of value",a.dtype)
print("shape of a",a.shape)
print("size of a",a.size)
print("dimention of a",a.ndim)
```
